    For this Data history review, I have decided to evaulate, analyze, and interpret the work of the organization "Our World In Data." This organization works hard to provide data from all parts of the world to facilitate data analysis about certain topics that affect all sides of the globe. OWID, by their definition, is "...produced as a collaborative effort between researchers at the University of Oxford, who are the scientific contributors of the website content, and the non-profit organization Global Change Data Lab, which owns, publishes, and maintains the website and the data tools." OWID has an extremely wide variety of data-driven research, but for this review specifically, I will be analyzing their data-induced research on life expectancy and it's growth throughout the world since 1543 (Table and Map Viewing) or 1770 (Line and Bar Graph Viewing) to 2023. This specific dataset and work was done by the founder himself, Max Roser and his co-director, Esteban Ortiz-Ospina, alongside a few of their lead and senior scientists, Saloni Dattani, Lucas Rodés-Guirao, and Hannah Ritchie. 


    The project is incredibly vast and filled with many outlets of research to display the point of view, which is, in short form, that life expectancy across the world has noticibly and visibly increased bya hefty margin since centuries prior, due to medical and technological advances, health and living standard advancements, and better quality of life. In a more complex sense, there is also the factors of way less child mortality over the years, that has largely contributed to the overall increase of life expectancy, and additionally, women tend to live longer than men, with a fluctuating gap:


    
    
    The interpretive text, alongside the four interactive visualization methods (Graph, Line, Bar, Table) do very well in communicating the point that the project is trying to convey. The world map and line graphs alone do well in summarizing the simple trend of increasing life expectancy, but the additional information takes things a step further by adding contextual intricacy to the trends, offering other points regarding demographical data.

    Investigating further into the data, we see that some of the data being used is measured in something called "Period Life Expectancy," which is, according to OWID, "...the number of years the average person born in a certain year would live if they experienced the same chances of dying at each age as people did that year." This method of measurement was used as a subgenre of life expectancy as a whole, which the interactive data had been using. The data used in the overarching set itself was sourced from  Riley (2005); Zijdeman et al. (2015); HMD (2025); UN WPP (2024), which is the regional expectancy data for the world from the range of years used, up until 1950, at which point the United Nations World Population Prospects (2024) was utilized. Even these sources are derived from hundreds of other combined sources. All of this data is computed into the interactive dataset, that highlights the movement of expectancy upward, as time goes on. The creators use the data very well, as they implement it all together in various forms that allow most to understand. And the data goes somewhat well with the extra texts giving more ideas to the project. However, although it is all nicely put together, there are still some casual complaints.





    To begin with, although the data is very neat and tidy, accompanying the great look and feel of the interation, a lot of the information and the text displaying more ideas come together quickly, and I imagine a casual reader may find it to be a but much at first glance. There are also extra articles that display different ideas relating to the overall topic of life expectancy, which some may not find immediately. The data and organization is great for researchers and analyists who want the most of what the data can offer, but for people who may have to read for an assignment, there will be a lot to consider and visualize. Still, the way the data is put together is very articulate and professional, and it can still cater to a wide variety of people. 

    The project’s argument is clearly communicated: it shows that life expectancy has risen dramatically worldwide and supports this claim with accessible charts, historical comparisons, and explanations of why these changes occurred. Its audience is a general public interested in global trends, including students, policymakers, and curious readers. The page meets that audience’s needs well by using plain language, interactive visualizations, and concise “Key Insights” that break down complex demographic data into understandable takeaways. It balances depth with clarity, making the evidence easier to follow without requiring technical knowledge or expertise.

    Overall, this project, and this website in general, provides great use of the data that is utilized, and because of that, can be a good way to analyze trends and historical events or images in a trustworthy way.
